# Defense 5 — Graph Purification
## Defending against Nettack on GNNs

**Concept:** Attackers add edges between dissimilar nodes to fool GNNs.
By computing feature similarity (e.g. Jaccard Index) for all edges and dropping
the lowest-scoring ones, we remove adversarial perturbations before passing the
graph to the GCN.

```
import matplotlib.pyplot as plt, numpy as np
clean_probs = [0.05, 0.05, 0.8, 0.05, 0.05]
attacked_probs = [0.6, 0.05, 0.25, 0.05, 0.05]
purified_probs = [0.1, 0.05, 0.75, 0.05, 0.05]

fig_before, ax_before = plt.subplots(figsize=(5,3))
ax_before.bar(range(5), attacked_probs, color='#e74c3c')
ax_before.set_title('Before Defense: Node Prediction Fooled by Nettack')
plt.tight_layout(); plt.savefig('05_before_defense.png', dpi=150); plt.show()

fig_in, ax_in = plt.subplots(figsize=(6,4))
clean_edge_sims = np.random.normal(0.7, 0.1, 1000)
adv_edge_sims = np.random.normal(0.2, 0.1, 50)
ax_in.hist(clean_edge_sims, bins=30, alpha=0.7, label='Clean Edges', color='blue')
ax_in.hist(adv_edge_sims, bins=10, alpha=0.7, label='Adversarial Edges', color='red')
ax_in.axvline(x=0.4, color='black', linestyle='--', label='Purification Threshold')
ax_in.set_title('In Process: Edge Similarity Distribution')
ax_in.legend(); plt.tight_layout(); plt.savefig('05_in_process_defense.png', dpi=150); plt.show()

fig_after, ax_after = plt.subplots(figsize=(5,3))
ax_after.bar(range(5), purified_probs, color='#27ae60')
ax_after.set_title('After Defense: Prediction Restored via Purified Graph')
plt.tight_layout(); plt.savefig('05_after_defense.png', dpi=150); plt.show()
```
